# **Data Acquisition with Scraping**
Data acquisition using Wikipedia.

```python
# import necessary libraries
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

## **Drivers**
F1 drivers — [Wikipedia](https://en.wikipedia.org/wiki/List_of_Formula_One_drivers)

```python
def scrape_drivers():
    url = "https://en.wikipedia.org/wiki/List_of_Formula_One_drivers"
    headers = {"User-Agent": "Mozilla/5.0 (educational-webscraping-lab)"}
    resp = requests.get(url, headers=headers)
    resp.raise_for_status()
    soup = BeautifulSoup(resp.text, "html.parser")

    tables = soup.find_all("table", {"class": "wikitable"})
    table = tables[1]  # we only need the second table

    headers = [th.get_text(strip=True) for th in table.find("tr").find_all("th")]

    data = []
    for row in table.find_all("tr")[1:]:
        cells = row.find_all("td")
        if len(cells) < 1:
            continue
        row_vals = [" ".join(cell.stripped_strings) for cell in cells[:len(headers)]]
        data.append(row_vals)

    df_final = pd.DataFrame(data, columns=headers)

    df_final.rename(columns={
        "Driver name": "driverName",
        "Nationality": "nationality",
        "Seasons competed": "seasons",
        "Drivers' Championships": "championships",
        "Race entries": "entries",
        "Race starts": "starts",
        "Pole positions": "poles",
        "Race wins": "wins",
        "Podiums": "podiums",
        "Fastest laps": "fastestLaps",
        "Points[a]": "points"
    }, inplace=True)

    return df_final

df_drivers = scrape_drivers()
df_drivers.to_csv("drivers_wiki.csv", index=False)
```

## **Constructors**
F1 constructors — [Wikipedia](https://en.wikipedia.org/wiki/List_of_Formula_One_constructors)

```python
def scrape_constructors():
    url = "https://en.wikipedia.org/wiki/List_of_Formula_One_constructors"
    headers = {"User-Agent": "Mozilla/5.0 (educational-webscraping-lab)"}
    resp = requests.get(url, headers=headers)
    resp.raise_for_status()
    soup = BeautifulSoup(resp.text, "html.parser")

    tables = soup.find_all("table", {"class": "wikitable"})
    
    dfs = []
    for table in tables[:2]:  # we need only the first two tables
        headers = [th.get_text(strip=True) for th in table.find("tr").find_all("th")]

        data = []
        for row in table.find_all("tr")[1:]:
            cells = row.find_all("td")
            if len(cells) < 1:
                continue
            row_vals = [" ".join(cell.stripped_strings) for cell in cells[:len(headers)]]
            data.append(row_vals)

        df = pd.DataFrame(data, columns=headers)
        dfs.append(df)

    df_final = pd.concat(dfs, ignore_index=True, sort=False)

    df_final.rename(columns={
        "Constructor": "constructor",
        "Engine": "engine",
        "Licensed in": "licensedIn",
        "Based in": "basedIn",
        "Seasons": "seasons",
        "Races Entered": "racesEntered",
        "Races Started": "racesStarted",
        "Drivers": "drivers",
        "Total Entries": "totalEntries",
        "Wins": "wins",
        "Points": "points",
        "Poles": "poles",
        "FL": "fastestLaps",
        "Podiums": "podiums",
        "WCC": "wcc",
        "WDC": "wdc",
        "Antecedent teams": "antecedentTeams"
    }, inplace=True)
    
    return df_final

df_constructors = scrape_constructors()
df_constructors.to_csv("constructors_wiki.csv", index=False)
```

## **Circuits**
F1 circuits — [Wikipedia](https://en.wikipedia.org/wiki/List_of_Formula_One_circuits)

```python
def scrape_circuits():
    url = "https://en.wikipedia.org/wiki/List_of_Formula_One_circuits"
    headers = {"User-Agent": "Mozilla/5.0 (educational-webscraping-lab)"}
    resp = requests.get(url, headers=headers)
    resp.raise_for_status()
    soup = BeautifulSoup(resp.text, "html.parser")

    tables = soup.find_all("table", {"class": "wikitable"})
    table = tables[1]  # we only need the second table

    headers = [th.get_text(strip=True) for th in table.find("tr").find_all("th")]

    data = []
    for row in table.find_all("tr")[1:]:
        cells = row.find_all("td")
        if len(cells) < 1:
            continue
        row_vals = [" ".join(cell.stripped_strings) for cell in cells[:len(headers)]]
        data.append(row_vals)

    df_final = pd.DataFrame(data, columns=headers)

    df_final.rename(columns={
        'Circuit': 'circuitName',
        'Map': 'map',
        'Type': 'type',
        'Direction': 'direction',
        'Location': 'city',
        'Country': 'country',
        'Last length used': 'length_km',
        'Turns': 'turns',
        'Grands Prix': 'grandPrix',
        'Season(s)': 'seasons',
        'Grands Prix held': 'grandPrixHeld'
    }, inplace=True)
    
    return df_final

df_circuits = scrape_circuits()
df_circuits.to_csv("circuits_wiki.csv", index=False)
```
