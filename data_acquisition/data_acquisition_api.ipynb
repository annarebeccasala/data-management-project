# **Data Acquisition with API**
Data acquisition using the Jolpica F1 API.

```python
# import necessary libraries
import requests
import pandas as pd
```

## **Results**
2025 F1 results — [API endpoint](https://api.jolpi.ca/ergast/f1/2025/results)

```python
def get_results():
    limit = 100  # max supported by the API
    offset = 0
    all_results = []

    while True:
        url = f"https://api.jolpi.ca/ergast/f1/2025/results.json?limit={limit}&offset={offset}"
        response = requests.get(url)
        response.raise_for_status()
        data = response.json()

        races = data["MRData"]["RaceTable"]["Races"]
        if not races:
            break  # terminated

        for race in races:
            for res in race["Results"]:
                all_results.append({
                    "round": race["round"],
                    "driverId": res["Driver"]["driverId"],
                    "constructorId": res["Constructor"]["constructorId"],
                    "grid": res["grid"],
                    "position": res["position"],
                    "points": res["points"],
                    "status": res["status"], 
                    "laps" : res["laps"],
                    "time": res.get("Time", {}).get("time"), #it could be null
                    "fastestLap": res.get("FastestLap", {}).get("Time", {}).get("time") #it could be null
                })

        offset += limit  # go to the next page

    df = pd.DataFrame(all_results)
    return df

df_results = get_results()
df_results.to_csv("results_api.csv", index=False)
```

## **Drivers**
2025 F1 drivers — [API endpoint](https://api.jolpi.ca/ergast/f1/2025/drivers)

```python
def get_drivers():
    limit = 100  # max supported by the API
    offset = 0
    all_results = []

    while True:
        url = f"https://api.jolpi.ca/ergast/f1/2025/drivers.json?limit={limit}&offset={offset}"
        response = requests.get(url)
        response.raise_for_status()
        data = response.json()

        drivers = data["MRData"]["DriverTable"]["Drivers"]
        if not drivers:
            break  # terminated

        for driver in drivers:
            all_results.append({
                "driverId": driver["driverId"],
                "number": driver.get("permanentNumber"), 
                "name": driver["givenName"],
                "surname" : driver["familyName"],
                "dateOfBirth": driver.get("dateOfBirth"),
            })

        offset += limit  # go to the next page

    df = pd.DataFrame(all_results)
    return df

df_drivers = get_drivers()
df_drivers.to_csv("drivers_api.csv", index=False)
```

## **Constructors**
2025 F1 constructors — [API endpoint](https://api.jolpi.ca/ergast/f1/2025/constructors)

```python
def get_constructors():
    limit = 100  # max supported by the API
    offset = 0
    all_results = []

    while True:
        url = f"https://api.jolpi.ca/ergast/f1/2025/constructors.json?limit={limit}&offset={offset}"
        response = requests.get(url)
        response.raise_for_status()
        data = response.json()

        constructors = data["MRData"]["ConstructorTable"]["Constructors"]
        if not constructors:
            break  # terminated

        for constructor in constructors:
            all_results.append({
                "constructorId": constructor["constructorId"],
                "constructorName": constructor["name"]
            })

        offset += limit  # go to the next page

    df = pd.DataFrame(all_results)
    return df

df_constructors = get_constructors()
df_constructors.to_csv("constructors_api.csv", index=False)
```

## **Races**
2025 F1 races — [API endpoint](https://api.jolpi.ca/ergast/f1/2025/races)

```python
def get_races():
    limit = 100  # max supported by the API
    offset = 0
    all_results = []

    while True:
        url = f"https://api.jolpi.ca/ergast/f1/2025/races.json?limit={limit}&offset={offset}"
        response = requests.get(url)
        response.raise_for_status()
        data = response.json()

        races = data["MRData"]["RaceTable"]["Races"]
        if not races:
            break  # terminated

        for race in races:
            sprint = race.get("Sprint", {})
            all_results.append({
                "round": race["round"],
                "raceName": race["raceName"],
                "date": race["date"],
                "time": race["time"],
                "sprintDate": sprint.get("date"), # it could be null
                "sprintTime": sprint.get("time"), # it could be null
                "circuitId": race["Circuit"]["circuitId"]
            })

        offset += limit  # go to the next page

    df = pd.DataFrame(all_results)
    return df

df_races = get_races()
df_races.to_csv("races_api.csv", index=False)
```
